# RAG Fine-tune Experiment Tracking
Uses MLflow + W&B for experiment_tracker and model_registry components.

```python
import mlflow
import wandb
from src.config.settings import settings

mlflow.set_tracking_uri(settings.mlflow_tracking_uri)
mlflow.set_experiment("rag-finetune-v2")
wandb.init(project="enterprise-rag-agent", name="notebook-run")

with mlflow.start_run():
    mlflow.log_param("base_model", "meta-llama/Llama-3.1-8B-Instruct")
    mlflow.log_param("method", "lora")
    mlflow.log_metric("eval_rouge_l", 0.41)
    mlflow.log_metric("eval_bertscore", 0.78)
    mlflow.sklearn.log_model(None, "placeholder")
    print("Logged to MLflow and W&B")
```
